#### Task 3.1: Tidy the data (5 points)

Produce a tidy data set that could be used to answer your research question using the `tidyverse`. Please ensure you have at least one quantitative (numeric) and one categorical variable:
- Create new variables (if needed)
- Transform the data into a tidy form (if needed)
- Remove irrelevant columns (if needed)
- Comment your code throughout
- Show the first 6 rows of the tidied data.

</div>

```r
# Each row represents one team's tournament appearance in a particular year.
# Each variable already has its own column, so reshaping is unnecessary.
basketball_tidy <- basketball_data |>
    # Keep regular-season wins, regular-season losses, and home-court status.
    select(reg_wins, reg_losses, first_game_at_home) |>
    # Convert home-court status into a categorical variable for classification.
    mutate(first_game_at_home = as_factor(first_game_at_home))

# Display the first 6 rows of the tidied data.
head(basketball_tidy)
```

```r
# Show the distribution of regular-season wins.
wins_distribution <- ggplot(basketball_tidy, aes(x = reg_wins)) + 
    geom_histogram(binwidth = 1) + 
    labs(title = "Distribution of regular-season wins", 
         x = "Regular-season wins (games)", 
         y = "Number of team tournament appearances") +
    theme(text = element_text(size = 12))

wins_distribution
```

```r
home_proportions <- basketball_tidy |>
    group_by(first_game_at_home) |> 
    summarize(count = n()) |>
    mutate(proportion = count/sum(count))

# Plot the proportion of appearances in each category. 
home_proportion_plot <- ggplot(home_proportions, 
                               aes(x = first_game_at_home, y = proportion)) + 
    geom_bar(stat = "identity") + 
    labs(title = "Home status of the first tournament game", 
        x = "First game at home (Y = yes, N = no)", 
        y = "Proportion of team tournament appearances") +
    theme(text = element_text(size = 12))

home_proportion_plot
```

Visualization 1 compares regular-season wins for team appearances with a first game at home (Y) and without one (N). The home-game group tends to have more wins, but the distributions overlap. This makes wins relevant as a possible predictor of home-game status while showing that wins alone may not fully distinguish the categories. Because the groups differ in size, the comparison focuses on the distributions’ horizontal positions and overlap rather than their bar heights.

```r
# Compare the distribution of wins between home-game categories.
wins_home_plot <- ggplot(basketball_tidy, aes(x = reg_wins)) +
    geom_histogram(binwidth = 1) +
    facet_grid(rows = vars(first_game_at_home)) +
    labs(title = "Wins by home status (Y = home, N = not home)",
         x = "Regular-season wins (games)",
         y = "Number of team tournament appearances") +
    theme(text = element_text(size = 12))

wins_home_plot
```

Visualization 2 compares regular-season losses between team appearances with a first game at home (Y) and without one (N). The home-game group tends to have fewer losses, although the distributions overlap. This suggests losses may help predict home-game status when used with wins in the classification model. Model evaluation is needed to assess how accurately these predictors classify new observations.

```r
# Compare the distribution of losses between home-game categories.
losses_home_plot <- ggplot(basketball_tidy, aes(x = reg_losses)) +
    geom_histogram(binwidth = 1) +
    facet_grid(rows = vars(first_game_at_home)) +
    labs(title = "Losses by home status (Y = home, N = not home)",
         x = "Regular-season losses (games)",
         y = "Number of team tournament appearances") +
    theme(text = element_text(size = 12))

losses_home_plot
```
